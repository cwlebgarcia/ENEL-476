### 1. Plot the electric and magnetic fields as a function of z at t= 0 in 3D.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# -----------------------------
# Constants (free space)
# -----------------------------
freq = 3e9                 # Hz
E0 = 1.0                   # V/m
c  = 3e8                   # m/s
eta0 = 377.0               # ohms

lam = c / freq
k = 2*np.pi / lam
omega = 2*np.pi * freq
t0 = 0.0

# z axis (2 wavelengths)
z = np.linspace(0, 2*lam, 400)
phase = omega*t0 - k*z

# Fields
E = E0 * np.cos(phase)
H = (E0/eta0) * np.cos(phase)

# -------- VISUAL SCALING (for H only) --------
H_vis_scale = 400
H_plot = H_vis_scale * H

# -----------------------------
# 3D plot
# -----------------------------
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Electric field line (oscillates in x)
ax.plot(E, np.zeros_like(z), z, color='r', linewidth=3)

# Magnetic field line (oscillates in y)
ax.plot(np.zeros_like(z), H_plot, z, color='b', linewidth=3)

# Reference z-axis
ax.plot([0,0], [0,0], [z.min(), z.max()], 'k--', linewidth=1)

ax.set_xlabel('z')
ax.set_ylabel('y')
ax.set_zlabel('x')
ax.set_title('Plane Wave in Free Space: E (red) and H (blue) vs z at t = 0')

ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)
ax.set_zlim(-0.3, 0.3)

# Legend
E_handle = mlines.Line2D([], [], color='r', label='E field (V/m)')
H_handle = mlines.Line2D([], [], color='b', label='H field (A/m) plotted ×300')
ax.legend(handles=[E_handle, H_handle], loc='upper right')

plt.show()
```

### 2. Plot the electric and magnetic fields as a function of time at z= 0 in 2D

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Variables (EDIT THESE)
# -----------------------------
E0 = 1.0            # V/m (amplitude)
alpha = 0         # Np/m (attenuation)  <-- doesn't affect z=0 but included
beta  = 62.87        # rad/m (phase constant) <-- doesn't affect z=0 but included
freq = 3e9          # Hz
omega = 2*np.pi*freq  # rad/s

eta_mag = 376.7    # |eta| in ohms (free space approx)
theta_eta = 0.0     # theta_eta in radians (phase of intrinsic impedance)

z0 = 0.0            # meters

# -----------------------------
# Time axis
# -----------------------------
T = 1/freq                       # period (s)
t = np.linspace(0, 3*T, 2000)    # show 3 cycles

# -----------------------------
# Eq (1) and Eq (3)
# -----------------------------
E = E0 * np.exp(-alpha*z0) * np.cos(omega*t - beta*z0)                         # Eq (1)
H = (E0/eta_mag) * np.exp(-alpha*z0) * np.cos(omega*t - beta*z0 - theta_eta) *300  # Eq (3)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(8,4.5))
plt.plot(t, E, label=r'$E(0,t)=E_0 e^{-\alpha z}\cos(\omega t-\beta z)$ at $z=0$')
plt.plot(t, H, label=r'$H(0,t)=\frac{E_0}{|\eta|}e^{-\alpha z}\cos(\omega t-\beta z-\theta_\eta)$ at $z=0$')

plt.xlabel('t (s)')
plt.ylabel('Field value')
plt.title('UPW at z = 0: E and H * 300 vs time')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

### 10. What impact does increasing the permittivity have on the wavelength?

Increasing permittivity would reduce wave speed since:
$$v = \frac{1}{\sqrt{\mu \varepsilon}}$$
Reducing wave speed would then results in decreasing wavelength:
$$ \lambda = \frac{v}{f}$$
Therefore increasing permittivity would reduce wavelength.

### 11. How does decreasing the frequency impact the wavelength?

Decreasing frequency would increase the wavelength:
$$ \lambda = \frac{v}{f}$$

### 12. What impact does increasing the permittivity have on the magnetic field strength?

For a uniform plane wave, the intrinsic impedance of a medium is:

$$
\eta = \sqrt{\frac{\mu}{\varepsilon}}
$$

Increasing the permittivity $\varepsilon$ (with $\mu$ constant) decreases the intrinsic impedance $\eta$.

The electric and magnetic fields are related by

$$
|H| = \frac{|E|}{|\eta|}
$$

Therefore, if the electric field magnitude $|E|$ is held constant, increasing the permittivity results in an increase in the magnetic field strength $|H|$.

### 13. How far does the wave travel before it undergoes 1$Np$ of attenuation when εr = 4 and σ= 0.1 S/m.

Based on pre-lab the attenuation constant is:

$$
\alpha = 9.39 \text{Np/m}
$$
And
$$
z = \frac{1}{\alpha}
$$

Therefore,

$$
z = \frac{1}{9.39} = 0.1065\ \text{m}
$$

**Answer:**  
The wave travels approximately 0.1065 m before undergoing 1 Np of attenuation.


### 14. Describe the behavior of the wave resulting from the sum of forward and backward traveling waves over time?

The consequence of summing forward and backward traveling waves is the formation of a standing wave, which results in fixed nodes and antinodes along the medium. At nodes, the field magnitude is always zero, while at antinodes it is maximized. This leads to non-uniform field distribution and no net power propagation, which can cause localized regions of high field strength and increased losses or voltage stress.
