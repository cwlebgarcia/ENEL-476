### 1. Plot the electric and magnetic fields as a function of z at t= 0 in 3D.

```python
import scipy
from scipy.constants import epsilon_0, mu_0, c
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import axes3d

E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
conductivity = 0 # S / m
# Wave travelling in the z direction

alpha = 0
beta = angular_frequency  * np.sqrt(mu_0 * epsilon_0)
eta = np.sqrt(mu_0 / epsilon_0)


N = int(1e4)
t = 0
z = np.linspace(0, z_max, N)

E_x = E_0 * np.exp(-alpha * z) * np.cos(angular_frequency * t - beta * z)
H_y = E_x / np.abs(eta)

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plotting E_x and H_y as functions of z
ax.plot3D(z, H_y, 0, label='H_y vs z', color='b')
ax.plot3D(z, 0, E_x, label='E_x vs z', color='r')

# Adding labels and title
ax.set_xlabel('z-axis')
ax.set_ylabel('H_y component')
ax.set_zlabel('E_x component')
ax.set_title('3D Plot of E_x and H_y as Functions of z')

# Show grid
ax.grid(True)
ax.legend()
plt.show()
```

### 3. Plot the magnitude of the electric and magnetic field phasors in 3D.

```python
import scipy
from scipy.constants import epsilon_0, mu_0, c
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import axes3d
import cmath

E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
conductivity = 0 # S / m
# Wave travelling in the z direction

alpha = 0
beta = angular_frequency  * np.sqrt(mu_0 * epsilon_0)
eta = np.sqrt(mu_0 / epsilon_0)
omega = complex(alpha, beta)

N = int(1e4)
t = 0
z = np.linspace(0, z_max, N)

E_x = E_0 * np.exp(-1 * omega * z)
H_y = E_x / np.abs(eta)
E_x = np.abs(E_x)
H_y = np.abs(H_y)

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plotting E_x and H_y as functions of z
ax.plot3D(z, H_y, 0, label='H_y vs z', color='b')
ax.plot3D(z, 0, E_x, label='E_x vs z', color='r')

# Adding labels and title
ax.set_xlabel('z-axis')
ax.set_ylabel('H_y component')
ax.set_zlabel('E_x component')
ax.set_title('3D Phasor Magnitude Plot of E_x and H_y as Functions of z')

# Show grid
ax.grid(True)
ax.legend()
plt.show()
```
