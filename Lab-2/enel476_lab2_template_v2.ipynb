### 1. Plot the electric and magnetic fields as a function of z at t= 0 in 3D.

```python
from scipy.constants import epsilon_0, mu_0, c
import numpy as np
import matplotlib.pyplot as plt

E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
conductivity = 0 # S / m
# Wave travelling in the z direction

alpha = 0
beta = angular_frequency  * np.sqrt(mu_0 * epsilon_0)
eta = np.sqrt(mu_0 / epsilon_0)

N = int(1e4)
t = 0
z = np.linspace(0, z_max, N)

E_x = E_0 * np.exp(-alpha * z) * np.cos(angular_frequency * t - beta * z)
H_y = E_x / np.abs(eta)

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plotting E_x and H_y as functions of z
ax.plot3D(0, H_y, z, label='H_y vs z', color='b')
ax.plot3D(E_x, 0, z, label='E_x vs z', color='r')

# Adding labels and title
ax.set_xlabel('E_x component')
ax.set_ylabel('H_y component')
ax.set_zlabel('z-axis')
ax.set_title('3D Plot of E_x and H_y as Functions of z')

# Show grid
ax.grid(True)
ax.legend()
plt.show()
```

### 4. Generate an animation of the fields for 0 ≤t≤1 ns in 3D.

```python
E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
conductivity = 0 # S / m

alpha = 0
beta = angular_frequency  * np.sqrt(mu_0 * epsilon_0)
eta = np.sqrt(mu_0 / epsilon_0)

N = 100
t = np.linspace(0, t_max, N)
z = np.linspace(0, z_max, N)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.set_xlim(-1, 1)
ax.set_ylim(-1/eta, 1/eta)
ax.set_zlim(0, z_max)

ax.set_xlabel("E_x (V/m)")
ax.set_ylabel("H_y (A/m)")
ax.set_zlabel("z")

E_line, = ax.plot([], [], [], 'b', label="Electric field")
H_line, = ax.plot([], [], [], 'r', label="Magnetic field")

ax.legend()

def update(frame):

    time = t[frame]

    E_x = E_0 * np.exp(-alpha * z) * np.cos(angular_frequency * time - beta * z)
    H_y = E_x / np.abs(eta)

    E_line.set_data(E_x, np.zeros_like(z))
    E_line.set_3d_properties(z)

    H_line.set_data(np.zeros_like(z), H_y)
    H_line.set_3d_properties(z)

    return E_line, H_line

ani = animation.FuncAnimation(fig, update, frames=len(t), interval=50)

plt.close()
HTML(ani.to_html5_video())
```

### 7. Wave traveling in +z-direction in a homogeneous medium with εr = 4,µr = 1, and σ= 0.1 S/m at 3 GHz.

```python
E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
relative_permittivity = 4
relative_permeability = 1
conductivity = 0.1 # S / m
# Wave travelling in the z direction

epsilon = epsilon_0 * relative_permittivity
mu = mu_0 * relative_permeability
alpha = angular_frequency * np.sqrt(mu * epsilon / 2 * 
                                    (np.sqrt(1 + np.power(conductivity / (angular_frequency * epsilon), 2) - 1))
                                    )
beta = angular_frequency * np.sqrt(mu * epsilon / 2 * 
                                    (np.sqrt(1 + np.power(conductivity / (angular_frequency * epsilon), 2) + 1))
                                    )
eta = np.sqrt(mu / epsilon) / np.power(1 + np.power(conductivity / (angular_frequency * epsilon), 2), 1/4)
theta_eta = 1/2 * np.atan(conductivity / (angular_frequency * epsilon))

N = 100
t = np.linspace(0, t_max, N)
z = z = np.linspace(0, z_max, N)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.set_xlim(-1, 1)
ax.set_ylim(-1/eta, 1/eta)
ax.set_zlim(0, z_max)

ax.set_xlabel("E_x (V/m)")
ax.set_ylabel("H_y (A/m)")
ax.set_zlabel("z")

E_line, = ax.plot([], [], [], 'b', label="Electric field")
H_line, = ax.plot([], [], [], 'r', label="Magnetic field")

ax.legend()

def update(frame):

    time = t[frame]

    E_x = E_0 * np.exp(-1 * alpha * z) * np.cos(angular_frequency * time - beta * z)
    H_y = E_0 * np.exp(-1 * alpha * z) * np.cos(angular_frequency * time - beta * z - theta_eta) / np.abs(eta)

    E_line.set_data(E_x, np.zeros_like(z))
    E_line.set_3d_properties(z)

    H_line.set_data(np.zeros_like(z), H_y)
    H_line.set_3d_properties(z)

    return E_line, H_line

ani = animation.FuncAnimation(fig, update, frames=len(t), interval=50)

plt.close()
HTML(ani.to_html5_video())
```

### 9. The sum of a forward-traveling wave with magnitude 1 and phase 0, and a backward-traveling wave with magnitude 0.5 and phase π/4 where both are traveling in free space and both have E-field oriented in the +x- direction.

```python
E_0 = 1 # V/m
phase_0 = 0 # rad
E_1 = 0.5 # V/m
phase_1 = np.pi / 4 # rad
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
relative_permittivity = 1
relative_permeability = 1
conductivity = 0 # S / m

epsilon = epsilon_0 * relative_permittivity
mu = mu_0 * relative_permeability
alpha = angular_frequency * np.sqrt(mu * epsilon / 2 * 
                                    (np.sqrt(1 + np.power(conductivity / (angular_frequency * epsilon), 2) - 1))
                                    )
beta = angular_frequency * np.sqrt(mu * epsilon / 2 * 
                                    (np.sqrt(1 + np.power(conductivity / (angular_frequency * epsilon), 2) + 1))
                                    )
eta = np.sqrt(mu / epsilon) / np.power(1 + np.power(conductivity / (angular_frequency * epsilon), 2), 1/4)
theta_eta = 1/2 * np.atan(conductivity / (angular_frequency * epsilon))

N = 100
t = np.linspace(0, t_max, N)
z = t * -c

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.set_xlim(-1.2*(E_0+E_1), 1.2*(E_0+E_1))
ax.set_ylim(-1.2*(E_0+E_1)/eta, 1.2*(E_0+E_1)/eta)
ax.set_zlim(-z_max, 0)

ax.set_xlabel("E_x (V/m)")
ax.set_ylabel("H_y (A/m)")
ax.set_zlabel("z")

E_line, = ax.plot([], [], [], 'b', label="Electric field")
H_line, = ax.plot([], [], [], 'r', label="Magnetic field")

ax.legend()

def update(frame):

    time = t[frame]

    E_x0 = E_0 * np.exp(-alpha*z) * np.cos(angular_frequency*time - beta*z - phase_0)
    H_y0 = E_0 * np.exp(-alpha*z) * np.cos(angular_frequency*time - beta*z - theta_eta - phase_0) / np.abs(eta)

    E_x1 = E_1 * np.exp(-alpha*z) * np.cos(-angular_frequency*time - beta*z - phase_1)
    H_y1 = E_1 * np.exp(-alpha*z) * np.cos(-angular_frequency*time - beta*z - theta_eta - phase_1) / np.abs(eta)

    E = E_x0 + E_x1
    H = H_y0 + H_y1

    E_line.set_data(E, np.zeros_like(z))
    E_line.set_3d_properties(z)

    H_line.set_data(np.zeros_like(z), H)
    H_line.set_3d_properties(z)

    return E_line, H_line

ani = animation.FuncAnimation(fig, update, frames=len(t), interval=50)

plt.close()
HTML(ani.to_html5_video())
```

### 10. What impact does increasing the permittivity have on the wavelength?

The velocity of the wave in the medium is determined by the equation

$$v = \frac{1}{\sqrt{\mu\epsilon}}$$

Which can be substituted into the equation

$$v = f\lambda$$

This can be rearranged into the equation

$$\lambda = \frac{f}{\sqrt{\mu\epsilon}}$$

Thus increasing the permittivity will have the effect of increasing the wavelength.

### 11. How does decreasing the frequency impact the wavelength?

Decreasing the freqeuncy has the effect of increasing the wavelength. This can easily be seen in the equation

$$c = f\lambda $$

Where c is a constant. Thus so maintain equality, if frequency is increased wavelength must increase.

### 12. What impact does increasing the permittivity have on the magnetic field strength?

Increasing the permittivity will have no effect on the magentic field strength as it is an electrical property. Changing the permeability will have an effect on the magnetic field strength, but the permittvity will not.

$$\alpha = \omega\sqrt{\frac{\mu\epsilon}{2}\left[\sqrt{1+\left[\frac{\sigma}{\omega\epsilon}\right]^2}-1\right]}$$

Assuming $\mu _r = 1$ and $f = 3GHz$

$$\alpha = 34.144 \frac{Np}{m}$$

$$ \therefore \delta=\alpha^{-1}=0.2905m$$
