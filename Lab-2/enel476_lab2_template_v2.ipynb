### 1. Plot the electric and magnetic fields as a function of z at t= 0 in 3D.

```python
from scipy.constants import epsilon_0, mu_0, c
import numpy as np
import matplotlib.pyplot as plt

E_0 = 1 # V/m
f = int(3e9) # GHz
angular_frequency = 2 * np.pi * f
t_max = 1e-9 # 1 ns
z_max = 30e-2 # m
conductivity = 0 # S / m
# Wave travelling in the z direction

alpha = 0
beta = angular_frequency  * np.sqrt(mu_0 * epsilon_0)
eta = np.sqrt(mu_0 / epsilon_0)

N = int(1e4)
t = 0
z = np.linspace(0, z_max, N)

E_x = E_0 * np.exp(-alpha * z) * np.cos(angular_frequency * t - beta * z)
H_y = E_x / np.abs(eta)

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plotting E_x and H_y as functions of z
ax.plot3D(z, H_y, 0, label='H_y vs z', color='b')
ax.plot3D(z, 0, E_x, label='E_x vs z', color='r')

# Adding labels and title
ax.set_xlabel('z-axis')
ax.set_ylabel('H_y component')
ax.set_zlabel('E_x component')
ax.set_title('3D Plot of E_x and H_y as Functions of z')

# Show grid
ax.grid(True)
ax.legend()
plt.show()
```

### 7. Wave traveling in +z-direction in a homogeneous medium with εr = 4,µr = 1, and σ= 0.1 S/m at 3 GHz.

### 10. What impact does increasing the permittivity have on the wavelength?

The velocity of the wave in the medium is determined by the equation

$$v = \frac{1}{\sqrt{\mu\epsilon}}$$

Which can be substituted into the equation

$$v = f\lambda$$

This can be rearranged into the equation

$$\lambda = \frac{f}{\sqrt{\mu\epsilon}}$$

Thus increasing the permittivity will have the effect of increasing the wavelength.

### 11. How does decreasing the frequency impact the wavelength?

Decreasing the freqeuncy has the effect of increasing the wavelength. This can easily be seen in the equation

$$c = f\lambda $$

Where c is a constant. Thus so maintain equality, if frequency is increased wavelength must increase.
